```python
# Adding term_freq (basic) instead of plain ids 
# so
# inverted_idx = { word : {doc_id1 : count(word) in doc_id1 ,doc_id2 : count(word) in doc_id2  } .. }

# so replace the set() with dict[doc_id] = freq(word in that doc_id)
    
```

```python
def term_freq_unnorm_idx(corpus_name,lang):
    inverted_index = {}
    file_lst = corpus_name.fileids()
    stopword = set(stopwords.words(lang))
    for id , doc in enumerate(file_lst):
        raw_words = corpus_name.words(fileid = doc)
        pro_words = [word.lower() for word in raw_words if word.lower() not in stopword]
        for word in pro_words:
            if word not in inverted_index : inverted_index[word] = {}
            inverted_index[word][id] = inverted_index[word].get(id,0) + 1
            
    return inverted_index , file_lst
        
raw_idx,corpus = term_freq_unnorm_idx(shakespeare,'english')       
```

```python
import math
def tf_(raw_counts):
    if raw_counts == 0 : return 0 
    return 1 + math.log10(raw_counts)

def idf_(word, inv_idx, file_lst):
    N = len(file_lst)
    df = len(inv_idx[word])
    return math.log10(N / df)
```

```python
def tf_idf_(inv_idx,word,doc_id,corpus_lst):
    if word not in inv_idx : return 0
    if doc_id not in inv_idx[word] : return 0
    raw_counts = inv_idx[word][doc_id] 
    if raw_counts == 0 : return 0
    tf = tf_(raw_counts)
    idf = idf_(word,inv_idx,corpus_lst)
    return tf*idf
    
    
    
    
```

```python
# Now create a function vectorize that 
# One imp thing is that the query word has to be in vocab word and cant be oov as tf-idf cant vectorize word that isnt present in the corpus
# First we vectorize the all the docs using the inv_idx and get a large dim sparse vector
# Same way we vectorize the query and then we perform the cosine similarity search
```

```python
def vectorize_docs_naive( doc_id , inv_idx , corpus):
    vocab = set(inv_idx.keys())
    doc_vec = {}
    for word in vocab :
        tf_idf = tf_idf_(inv_idx , word , doc_id , corpus )
        doc_vec[word] = tf_idf
    return doc_vec
        
        
def vectorize_all_docs(inv_idx , corpus):
    all_docs_vector = dict()
    for doc_id in range(len(corpus)):
        all_docs_vector[doc_id] = vectorize_docs_naive(doc_id , inv_idx , corpus)
        
    return all_docs_vector
```

```python
def vectorize_query_(query,inv_idx,corpus):
    pro_query = query.lower().split()
    query_vec = {}
    vocab = set(inv_idx.keys())
    for word in vocab :
        raw_counts = pro_query.count(word)
        if raw_counts == 0 : 
            query_vec[word] = 0
            continue
        tf = tf_(raw_counts)
        idf = idf_(word,inv_idx,corpus)
        query_vec[word] = tf * idf
    return query_vec


        
        
```

```python

def search_vector(query_vec, all_docs_vector):
    query_mag = math.sqrt(sum(val ** 2 for val in query_vec.values()))
    scores = {}
    for doc_id, doc_vec in all_docs_vector.items():
        dot_prod = 0
        doc_mag_sq = 0                                    
        for word, val in doc_vec.items():
            val_in_query = query_vec.get(word, 0)
            dot_prod += val_in_query * val               
            doc_mag_sq += val ** 2                       
        doc_mag = math.sqrt(doc_mag_sq)
        if query_mag == 0 or doc_mag == 0:               
            scores[doc_id] = 0
        else:
            scores[doc_id] = dot_prod / (query_mag * doc_mag)
    
    return scores
```

```python
def rank_results(scores, corpus, k=10):
   
    ranked = sorted(scores.items(), key=lambda x: x[1], reverse=True)
    ranked_named = {corpus[doc_id]: score for doc_id, score in ranked[:k]}
    
    return ranked_named
```
